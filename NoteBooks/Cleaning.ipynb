### Detecting outliers

### Detecting outliers for column "Tenure"

```python
df['tenure'].value_counts()
```

```python
tenure = df['tenure']

# 2. Mathematical Detection (IQR Method)
Q1 = tenure.quantile(0.25)
Q3 = tenure.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = df[(tenure < lower_bound) | (tenure > upper_bound)]

print("--- Tenure IQR Outlier Detection ---")
print(f"25th Percentile (Q1): {Q1}")
print(f"75th Percentile (Q3): {Q3}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
print(f"Minimum Value in Data: {tenure.min()}")
print(f"Maximum Value in Data: {tenure.max()}")
print(f"Total Outliers Found: {len(outliers_iqr)}")
```

```python
# 3. Visual Detection (Boxplot)
plt.figure(figsize=(8, 4))
sns.boxplot(x=tenure, color='skyblue')
plt.title('Boxplot of Tenure')
plt.xlabel('Tenure (Months)')
plt.show()
```

### Detecting outliers for column "MonthlyCharges"

```python
df['MonthlyCharges'].min()
```

```python
df['MonthlyCharges'].max()
```

```python
monthly_charges = df['MonthlyCharges']
Q1 = monthly_charges.quantile(0.25)
Q3 = monthly_charges.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify rows outside the upper and lower bounds
outliers_iqr = df[(monthly_charges < lower_bound) | (monthly_charges > upper_bound)]

print("--- MonthlyCharges IQR Outlier Detection ---")
print(f"25th Percentile (Q1): {Q1}")
print(f"75th Percentile (Q3): {Q3}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
print(f"Minimum Value in Data: {monthly_charges.min()}")
print(f"Maximum Value in Data: {monthly_charges.max()}")
print(f"Total Outliers Found: {len(outliers_iqr)}")
```

```python
# 3. Visual Detection (Boxplot)
plt.figure(figsize=(8, 4))
sns.boxplot(x=monthly_charges, color='coral')
plt.title('Boxplot of MonthlyCharges')
plt.xlabel('Monthly Charges ($)')
plt.show()
```
